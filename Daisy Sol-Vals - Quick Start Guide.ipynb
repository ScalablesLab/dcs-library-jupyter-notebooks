# Daisy Sol-Vals - Quick Start Guide
## A guide for Daisy Sol-Val and Daisy Sol-Val XL

By the end of this notebook you will know how to connect your **Daisy Sol-Val** or **Daisy Sol-Val XL**, initialize it, and open and close solenoid valves.

Both modules are 8-channel solenoid manifolds — each valve can be controlled independently. 

## Initialize

- Always call `sol.initialize()` once after connecting before actuating any valves. This configures

## Open valves

- Use `sol.turn_on_valves({valve_numbers})` to open one or more valves. Pass a set of 1-based valve indices:
>    `{1}` — open valve 1 only \
>    `{1, 2, 4}` — open valves 1, 2, and 4 simultaneously

## Close valves

- Use `sol.turn_off_valves()` to de-energize all valves.

## Specifications

| Parameter | Sol-Val | Sol-Val XL |
|---|---|---|
| Channels | 8 (on 1 common port) | 8×1 or two 4×1 |
| Recommended max flow | ~100 ml/min | ~400 ml/min |
| Switching time | 20 ms | 30 ms |
| Internal volume | 9 µl | 21 µl |
| Recommended tubing | PTFE | PTFE (larger ID, e.g. 1/16" ID) |

## Valve numbering

Both modules support up to 8 solenoid valves and are addressed by their valve number.

Pass indices in a **set** (curly braces `{}`). Multiple valves can be opened at the same time. Plug unused valves with a stopper fitting.

### Sol-Val XL — 4×2 configuration

The Sol-Val XL can also be configured as **two independent 4×1 manifolds** (two separate common ports, each with 4 valves) instead of a single 8×1. In this configuration, valves 1–4 share one common port and valves 5–8 share a second common port — useful when you need two completely independent flow paths. 

| | **Valve Numbers**|
|---|---|
| **Common Port #1** | **1, 2, 3, 4**| 
| **Common Port #2** | **5, 6, 7, 8**|

## Import
The Daisy Python Library is already loaded into the Daisy Controller Software, all you have to do is import any external libraries you need. Here we are importing time to allow for pauses before opening or closing valves by using `time.sleep()`.

```python
import time
```

## Print instruments found
Print the instruments found to easily verify the index of the Daisy you will be using.

## Set your instrument variable name
Change `INST_INDEX` to the position of the Daisy (Sol Val) you want to use.

```python
# Set which instrument you are calibrating
INST_INDEX = 0   # 0 if first instrument or change to match your Sol Val's position

sol = self.my_interface.groups[0].inst[INST_INDEX]
print(f"Using: {sol.name}")
```

## Open and close valves

Change the set in `turn_on_valves()` to the valve indices you want to open.

```python
# Open valve 1
sol.turn_on_valves({1})
print("Valve 1 open")
```

```python
# Close all valves
sol.turn_off_valves()
print("All valves closed")
```

## Example valve operations

```python
# Open valve 1 for 2 seconds then close it
sol.turn_on_valves({1})
print("Valve 1 open")
time.sleep(2)
sol.turn_off_valves()
print("All valves closed")
```

```python
# Open valves 1, 3, and 5 simultaneously
sol.turn_on_valves({1, 3, 5})
print("Valves 1, 3, 5 open")
time.sleep(2)
sol.turn_off_valves()
print("All valves closed")
```

```python
# Step through valves 1, 2, 3 one at a time
for index in [1, 2, 3]:
    sol.turn_on_valves({index})
    print(f"Valve index {index} open")
    time.sleep(1)

sol.turn_off_valves()
print("All valves closed")
```
