# Daisy Process - Quickstart Guide

By the end of this notebook you will know how to connect your **Daisy Process** and take readings from a connected Hamilton ARC probe.

The Daisy Process is compatible with the full range of **Hamilton ARC probes:** pH, ORP, dissolved oxygen (DO), dissolved CO₂, conductivity, viable cell density, and total cell density.

## Channels

- A single Daisy Process supports up to **4 probe channels** (1–4). If you have a probe connected to channel 1, use `channel=1` or simply `1` for the argument in functions for the sensor.

## Identify your probe

- Use `sensor.get_probe_id(1)` to read the **probe ID** and **serial number** from a probe connected in channel 1.

## Read sensor parameters

- Use `sensor.measure_parameters(1)` to take a reading from a connected probe.
- The meaning of each value depends on the probe type connected
> For example, a pH probe measures and reports:
  `(pH, mV, T)`

## Compatible Hamilton ARC Probes

All Hamilton ARC probes connect to the Daisy Process via a **VP8 cable**.

---

#### pH Probe — Polilyte Plus Arc
Measures **pH, millivolts (mV), and temperature (°C)**. The transmitter is built into the probe head and calibration is stored onboard.

#### ORP Probes — Polilyte Plus ORP Arc
Measures **oxidation-reduction potential (mV) and temperature (°C)**.

#### Dissolved Oxygen Probes — VisiFerm/VisiTrace DO Arc
Measures **dissolved oxygen (DO) and temperature (°C)** using optical luminescence quenching.

#### Dissolved CO₂ Probes — CO₂NTROL Arc
Measures **dissolved CO₂ and temperature (°C)** using a solid-state mid-IR optical (NDIR) measurement.

#### Conductivity Probes — Conducell 4UxF Arc
Measures **conductivity and temperature (°C)**.

#### Viable Cell Density — Incyte Arc
Measures **viable cell density (VCD)** using electrical permittivity (capacitance). Only cells with intact membranes polarize in the applied electric field, so dead cells, debris, and aggregates are invisible to the measurement — the reading reflects only the living, metabolically active fraction of the culture.

#### Total Cell Density — Dencytee Arc
Measures **total cell density and turbidity** using optical transflectance, a combination of transmitted and reflected light at 860 nm.

## Import
The Daisy Python Library is already loaded into the Daisy Controller Software, all you have to do is import any external libraries you need. Here we are importing time to allow for pauses between readings by using `time.sleep()`.

```python
import time
```

## Print instruments found
Print the instruments found to easily verify the index of the Daisy you will be using.

```python
print("Instruments found:")
for position, (code, inst) in enumerate(zip(self.my_interface.instruments_list[0], self.my_interface.groups[0].inst)):
    print(f"  [{position}]  {code} — {inst.name}")
```

## Set your instrument variable name
Change `INST_INDEX` to the position of the Daisy (Process) you want to use.

```python
# Set which instrument you are using
INST_INDEX = 0   # 0 if first instrument, or change to match your sensor's position

sensor = self.my_interface.groups[0].inst[INST_INDEX]
print(f"Using: {sensor.name}")
```

## Identify your probe

Run the cell below to read the probe ID and serial number from your connected probe. This is useful for record-keeping and confirming the correct probe is connected.

```python
# Change CHANNEL to match the channel your probe is connected to (1–4)
CHANNEL = 2

probe_id, serial_no = sensor.get_probe_id(channel=CHANNEL)
print(f"Channel {CHANNEL} probe ID: {probe_id}")
print(f"Channel {CHANNEL} serial number: {serial_no}")
```

## Take a single reading

Submerge your probe in the liquid and allow a few seconds for it to stabilize, then run the cell below.

For a pH meter, `measure_parameters()` returns three values: `(pH, mV, T)`

If you happen to mislabel your channel, you will be returned values like -100 and -99.99.

```python
# Change CHANNEL to match the channel your probe is connected to (1–4)
CHANNEL = 1

pH, mV, T = sensor.measure_parameters(CHANNEL)
print(f"pH:                  {pH:.3f}")
print(f"Millivolts:          {mV:.3f} mV")
print(f"Temperature:         {T:.1f} °C")
```

## Take multiple readings

The cell below takes a reading every `INTERVAL_S` seconds for `NUM_READINGS` readings and prints each result. Consider your application, when setting seconds between reads. For example, if you are measuring your solution with a pH meter after adding in acid or base allow your solution to mix before taking a reading.

```python
CHANNEL      = 2     # channel your probe is connected to (1–4)
INTERVAL_S   = 10    # seconds between readings
NUM_READINGS = 3     # number of readings to take

print(f"Taking {NUM_READINGS} readings every {INTERVAL_S} s from channel {CHANNEL}...\n")
print(f"{'#':<5} {'Value 1':>10} {'mV':>10} {'Temp (°C)':>12}")
print("-" * 40)

for i in range(1, NUM_READINGS + 1):
    pH, mV, T = sensor.measure_parameters(channel=CHANNEL)
    print(f"{i:<5} {pH:>10.3f} {mV:>10.3f} {T:>12.1f}")
    if i < NUM_READINGS:
        time.sleep(INTERVAL_S)

print("\nReadings complete.")
```
